```python
weather['DATE'] = pd.to_datetime(weather['DATE'])
weather = weather.drop(columns=['STATION', 'NAME'])
weather = weather.rename(columns={'DATE': 'Date', 'PRCP': 'Precipitation'})
weather.head()
```

```python
rainy_days = (weather['Precipitation'] > 0).sum()
total_days = len(weather)
rain_pct = rainy_days / total_days * 100

print(f"Iš viso dienų: {total_days}")
print(f"Lietingų dienų: {rainy_days}")
print(f"Sausų dienų: {total_days - rainy_days}")
print(f"Lietingų dienų procentas: {rain_pct:.1f}%")
```

```python
daily = df[df['Time'].isin([pd.to_datetime('09:30').time(), pd.to_datetime('16:30').time()])].copy()
daily['Date'] = daily['DateTime'].dt.date
print(f"Eilučių po filtravimo: {len(daily)}")
daily.head(10)
```

```python
daily_wide = daily.pivot(index='Date', columns='Time', values='Close')
daily_wide.columns = ['ORB_Close', 'EOD_Close']
daily_wide = daily_wide.reset_index()
daily_wide['Day_Move'] = daily_wide['EOD_Close'] - daily_wide['ORB_Close']
daily_wide.head(10)
```

```python
daily_wide['Date'] = pd.to_datetime(daily_wide['Date'])
print("daily_wide Date type:", daily_wide['Date'].dtype)
print("weather Date type:", weather['Date'].dtype)
```

```python
merged = daily_wide.merge(weather, on='Date', how='inner')
print(f"Eilučių skaičius: {len(merged)}")
merged.head(10)
```

```python
merged['Rainy'] = merged['Precipitation'] > 0
merged['Direction'] = merged['Rainy'].apply(lambda x: 'Short' if x else 'Long')
merged['Weather_PnL'] = merged.apply(
    lambda row: row['Day_Move'] if row['Direction'] == 'Long' else -row['Day_Move'],
    axis=1
)
merged.head(10)
```

```python
total_pnl = merged['Weather_PnL'].sum()
avg_pnl = merged['Weather_PnL'].mean()
win_rate = (merged['Weather_PnL'] > 0).sum() / len(merged) * 100
total_days = len(merged)

print(f"Iš viso dienų: {total_days}")
print(f"Bendras PnL: {total_pnl:.1f} punktų")
print(f"Vidutinis PnL per dieną: {avg_pnl:.2f} punktų")
print(f"Win rate: {win_rate:.1f}%")
```

```python
# Benchmark 1: Visada Long
always_long_pnl = merged['Day_Move'].sum()
always_long_avg = merged['Day_Move'].mean()
always_long_win = (merged['Day_Move'] > 0).sum() / len(merged) * 100

# Benchmark 2: Visada Short
always_short_pnl = -merged['Day_Move'].sum()
always_short_avg = -merged['Day_Move'].mean()
always_short_win = (merged['Day_Move'] < 0).sum() / len(merged) * 100

print("=" * 50)
print(f"{'Strategy':<20} {'Total PnL':>10} {'Avg/day':>10} {'Win %':>8}")
print("=" * 50)
print(f"{'Weather-based':<20} {total_pnl:>10.1f} {avg_pnl:>10.2f} {win_rate:>7.1f}%")
print(f"{'Always Long':<20} {always_long_pnl:>10.1f} {always_long_avg:>10.2f} {always_long_win:>7.1f}%")
print(f"{'Always Short':<20} {always_short_pnl:>10.1f} {always_short_avg:>10.2f} {always_short_win:>7.1f}%")
print("=" * 50)
```

```python
# Atskiriam lietingos ir nelietingos dienos
rainy_days = merged[merged['Rainy'] == True]
dry_days = merged[merged['Rainy'] == False]

# Skaičiuojam kiekvienos grupės statistikas
print(f"{'Grupė':<25} {'Dienų':>8} {'Avg Move':>12} {'Win %':>8}")
print("=" * 55)

# Lietingos dienos — mes Short, tai laimim kai rinka krenta
rainy_short_win = (rainy_days['Day_Move'] < 0).sum() / len(rainy_days) * 100
print(f"{'Rainy (Short)':<25} {len(rainy_days):>8} {rainy_days['Day_Move'].mean():>12.2f} {rainy_short_win:>7.1f}%")

# Nelietingos dienos — mes Long, tai laimim kai rinka kyla
dry_long_win = (dry_days['Day_Move'] > 0).sum() / len(dry_days) * 100
print(f"{'Dry (Long)':<25} {len(dry_days):>8} {dry_days['Day_Move'].mean():>12.2f} {dry_long_win:>7.1f}%")
```

```python
# Sukuriam "ankstesnės dienos" lietingumo stulpelį
merged['Prev_Rainy'] = merged['Rainy'].shift(1)

# Numetam pirmą eilutę (ten Prev_Rainy bus NaN)
transitions = merged.dropna(subset=['Prev_Rainy'])

# Skaičiuojam perėjimų tikimybes
transition_matrix = pd.crosstab(
    transitions['Prev_Rainy'], 
    transitions['Rainy'], 
    normalize='index'
)

print("Markovo perėjimų matrica:")
print("(eilutė = vakar, stulpelis = šiandien)")
print()
print(transition_matrix.round(3))
```

```python
import numpy as np

# Parametrai
n_simulations = 10000
n_days = len(merged)

# Markovo matrica kaip paprasti kintamieji
p_rain_after_dry = transition_matrix.loc[False, True]   # 0.322
p_rain_after_rain = transition_matrix.loc[True, True]   # 0.435

# Tikrasis rinkos judesys kiekvieną dieną (nekeisime)
real_moves = merged['Day_Move'].values

# Vieta simuliacijų rezultatams
sim_results = np.zeros(n_simulations)

# Monte Carlo ciklas
np.random.seed(42)  # kad rezultatas būtų atkartojamas

for sim in range(n_simulations):
    # Generuojam sintetinę lietaus istoriją pagal Markovo matricą
    synthetic_rain = np.zeros(n_days, dtype=bool)
    synthetic_rain[0] = np.random.random() < 0.361  # pirma diena pagal bendrą lietaus dažnį
    
    for day in range(1, n_days):
        if synthetic_rain[day-1]:  # vakar lijo
            synthetic_rain[day] = np.random.random() < p_rain_after_rain
        else:  # vakar nelijo
            synthetic_rain[day] = np.random.random() < p_rain_after_dry
    
    # Pritaikom strategiją: jei sintetinis lietus -> Short, kitaip -> Long
    sim_pnl = np.where(synthetic_rain, -real_moves, real_moves).sum()
    sim_results[sim] = sim_pnl

print(f"Monte Carlo baigta: {n_simulations} simuliacijų")
print(f"Vidutinis PnL: {sim_results.mean():.1f}")
print(f"Standartinis nuokrypis: {sim_results.std():.1f}")
print(f"Tavo realus PnL: {total_pnl:.1f}")
print(f"Tavo rezultato percentilis: {(sim_results < total_pnl).sum() / n_simulations * 100:.1f}%")
```

```python
import numpy as np

# Pasiruoškim Markovo matricos parametrus (jeigu dar neturi)
merged_clean['Prev_Rainy'] = merged_clean['Rainy'].shift(1)
transitions = merged_clean.dropna(subset=['Prev_Rainy'])
transition_matrix = pd.crosstab(
    transitions['Prev_Rainy'], 
    transitions['Rainy'], 
    normalize='index'
)
p_rain_after_dry = transition_matrix.loc[False, True]
p_rain_after_rain = transition_matrix.loc[True, True]

# Parametrai
n_simulations = 1000  # mažiau nei 10k, nes saugome visas kreives — užtenka
n_days = len(merged_clean)
real_moves = merged_clean['Day_Move'].values

# Vieta visoms kumuliatyvioms kreivėms
sim_curves = np.zeros((n_simulations, n_days))

np.random.seed(42)

for sim in range(n_simulations):
    synthetic_rain = np.zeros(n_days, dtype=bool)
    synthetic_rain[0] = np.random.random() < 0.361
    
    for day in range(1, n_days):
        if synthetic_rain[day-1]:
            synthetic_rain[day] = np.random.random() < p_rain_after_rain
        else:
            synthetic_rain[day] = np.random.random() < p_rain_after_dry
    
    daily_pnl = np.where(synthetic_rain, -real_moves, real_moves)
    sim_curves[sim] = np.cumsum(daily_pnl)

# Reali strategijos kumuliatyvi kreivė
real_curve = merged_clean['Weather_PnL'].cumsum().values

print(f"Simuliacijos baigtos: {n_simulations} kreivės, kiekvienoje {n_days} dienos")
print(f"Realios strategijos galutinis PnL: {real_curve[-1]:.1f}")
```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Apskaičiuojam percentilius per visas simuliacijas
p5 = np.percentile(sim_curves, 5, axis=0)
p25 = np.percentile(sim_curves, 25, axis=0)
p50 = np.percentile(sim_curves, 50, axis=0)
p75 = np.percentile(sim_curves, 75, axis=0)
p95 = np.percentile(sim_curves, 95, axis=0)

# Datos X ašiai
dates = merged_clean['Date'].values

# Sukuriame grafiką su tamsia tema
fig, ax = plt.subplots(figsize=(16, 8), facecolor='#0d1b2a')
ax.set_facecolor('#0d1b2a')

# 5-95 percentilio juosta (šviesiausia)
ax.fill_between(dates, p5, p95, color='#1b4965', alpha=0.6, label='5th-95th percentile')

# 25-75 percentilio juosta (tamsiau)
ax.fill_between(dates, p25, p75, color='#2a9d8f', alpha=0.5, label='25th-75th percentile')

# Mediana (punktyrinė)
ax.plot(dates, p50, color='#a8dadc', linestyle='--', linewidth=1.5, label='Median path', alpha=0.8)

# Reali strategija (balta, ryški)
ax.plot(dates, real_curve, color='white', linewidth=2.5, label='Actual backtest')

# Anotacijos: peak ir current
peak_idx = np.argmax(real_curve)
peak_value = real_curve[peak_idx]
peak_date = dates[peak_idx]

current_value = real_curve[-1]
current_date = dates[-1]

# Drawdown nuo peak iki dabar
drawdown = (current_value - peak_value) / peak_value * 100 if peak_value > 0 else 0

# Peak taškas
ax.scatter([peak_date], [peak_value], color='#06d6a0', s=120, zorder=5, edgecolor='white', linewidth=1.5)
ax.annotate(f'Peak: +{peak_value:.0f} pts',
            xy=(peak_date, peak_value), xytext=(15, 15),
            textcoords='offset points', color='#06d6a0', fontsize=11, fontweight='bold')

# Current taškas
ax.scatter([current_date], [current_value], color='#f4a261', s=120, zorder=5, edgecolor='white', linewidth=1.5)
ax.annotate(f'Now: +{current_value:.0f} pts\n(DD: {drawdown:.1f}%)',
            xy=(current_date, current_value), xytext=(-120, 15),
            textcoords='offset points', color='#f4a261', fontsize=11, fontweight='bold')

# Stiliavimas
ax.set_title('MONTE CARLO — 1,000 Markov Chain Paths vs Actual Strategy', 
             color='white', fontsize=15, fontweight='bold', pad=20)
ax.set_xlabel('Date', color='white', fontsize=11, style='italic')
ax.set_ylabel('Cumulative PnL (points)', color='white', fontsize=11)
ax.tick_params(colors='white', labelsize=10)
ax.grid(True, alpha=0.15, color='white')
ax.axhline(y=0, color='white', linestyle=':', alpha=0.3, linewidth=0.8)

# Legenda
legend = ax.legend(loc='upper left', framealpha=0.3, facecolor='#0d1b2a', edgecolor='white')
for text in legend.get_texts():
    text.set_color('white')

# Apatinė pastaba
fig.text(0.5, 0.02, 
         'Each path simulates an alternative rainfall history using the same Markov transition probabilities — different days are rainy, but the underlying market moves remain real.',
         ha='center', color='#a8dadc', fontsize=9, style='italic')

# Spacing
plt.tight_layout()
plt.subplots_adjust(bottom=0.12)

# Išsaugom kaip PNG
plt.savefig('monte_carlo_paths.png', dpi=150, bbox_inches='tight', facecolor='#0d1b2a')

plt.show()

print("Grafikas išsaugotas: monte_carlo_paths.png")
```

```python
# Apibrėžiame "geras pavyzdys" kriterijus:
# Diena, kurioje judesys yra pasiskirstęs per visą dieną,
# o ne sukoncentruotas paskutinėse 2 žvakėse

def find_clean_examples(merged_clean, df):
    candidates = []
    
    for idx, row in merged_clean.iterrows():
        date_obj = row['Date'].date()
        is_rainy = row['Rainy']
        day_move = row['Day_Move']
        
        # Pradinis filtras
        if is_rainy and day_move > -50:
            continue
        if not is_rainy and day_move < 50:
            continue
        
        # Paimam visas tos dienos žvakes
        day = df[df['DateTime'].dt.date == date_obj].copy()
        day = day[
            (day['Time'] >= pd.to_datetime('09:30').time()) & 
            (day['Time'] <= pd.to_datetime('16:30').time())
        ]
        
        if len(day) < 25:
            continue
        
        orb_close = day.iloc[0]['Close']
        eod_close = day.iloc[-1]['Close']
        
        # 14:00 kainos pozicija — daugiau nei pusė judesio iki 14:00
        mid_day_close = day[day['Time'] == pd.to_datetime('14:00').time()]['Close']
        if len(mid_day_close) == 0:
            continue
        mid_day_close = mid_day_close.iloc[0]
        
        # Kiek procentų dienos judesio jau įvyko iki 14:00
        progress_at_14 = (mid_day_close - orb_close) / (eod_close - orb_close) if (eod_close - orb_close) != 0 else 0
        
        # Paskutinės žvakės dydis (16:30) lyginant su visa diena
        last_candle_size = abs(day.iloc[-1]['Close'] - day.iloc[-1]['Open'])
        total_move = abs(day_move)
        last_candle_ratio = last_candle_size / total_move if total_move > 0 else 0
        
        candidates.append({
            'date': date_obj,
            'rainy': is_rainy,
            'day_move': day_move,
            'progress_at_14': progress_at_14,
            'last_candle_ratio': last_candle_ratio
        })
    
    return pd.DataFrame(candidates)

candidates_df = find_clean_examples(merged_clean, df)

# Rinkimasis: bent 60% judesio iki 14:00 ir paskutinė žvakė ne daugiau 25% bendro judesio
clean_short = candidates_df[
    (candidates_df['rainy'] == True) & 
    (candidates_df['progress_at_14'] >= 0.6) & 
    (candidates_df['last_candle_ratio'] <= 0.25) &
    (candidates_df['day_move'] < -80)
].sort_values('day_move').head(10)

clean_long = candidates_df[
    (candidates_df['rainy'] == False) & 
    (candidates_df['progress_at_14'] >= 0.6) & 
    (candidates_df['last_candle_ratio'] <= 0.25) &
    (candidates_df['day_move'] > 80)
].sort_values('day_move', ascending=False).head(10)

print("TOP 10 'švarių' SHORT dienų (lietingos):")
print(clean_short[['date', 'day_move', 'progress_at_14', 'last_candle_ratio']].to_string(index=False))
print()
print("TOP 10 'švarių' LONG dienų (sausos):")
print(clean_long[['date', 'day_move', 'progress_at_14', 'last_candle_ratio']].to_string(index=False))
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Užtikrinam, kad matrica yra atminty
if 'transition_matrix' not in dir():
    merged_clean['Prev_Rainy'] = merged_clean['Rainy'].shift(1)
    transitions = merged_clean.dropna(subset=['Prev_Rainy'])
    transition_matrix = pd.crosstab(
        transitions['Prev_Rainy'], 
        transitions['Rainy'], 
        normalize='index'
    )

# Vizualizacija
fig, ax = plt.subplots(figsize=(10, 7), facecolor='#0d1b2a')
ax.set_facecolor('#0d1b2a')

matrix_values = transition_matrix.values
labels = ['Dry', 'Rainy']

# Custom colormap — tamsus į šviesų teal/cyan
from matplotlib.colors import LinearSegmentedColormap
colors_list = ['#0d1b2a', '#1b4965', '#2a9d8f', '#06d6a0']
custom_cmap = LinearSegmentedColormap.from_list('custom_teal', colors_list, N=256)

im = ax.imshow(matrix_values, cmap=custom_cmap, aspect='auto', vmin=0, vmax=1)

# Spalvinis bar'as
cbar = plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
cbar.set_label('Probability', color='white', fontsize=11)
cbar.ax.tick_params(colors='white')
cbar.outline.set_edgecolor('white')

# Tekstas ląstelėse
for i in range(2):
    for j in range(2):
        value = matrix_values[i, j]
        # Spalva priklauso nuo fono
        text_color = 'white' if value < 0.55 else '#0d1b2a'
        ax.text(j, i, f'{value:.1%}',
                ha='center', va='center',
                color=text_color, fontsize=22, fontweight='bold')
        
        # Mažesnis tekstas su dienų skaičiumi
        n_days_in_state = transitions[transitions['Prev_Rainy'] == bool(i)]
        n_total = len(n_days_in_state)
        n_count = int(value * n_total)
        ax.text(j, i + 0.35, f'(n = {n_count} days)',
                ha='center', va='center',
                color=text_color, fontsize=10, style='italic')

# Ašys
ax.set_xticks([0, 1])
ax.set_yticks([0, 1])
ax.set_xticklabels(['Today: DRY', 'Today: RAINY'], color='white', fontsize=12, fontweight='bold')
ax.set_yticklabels(['Yesterday: DRY', 'Yesterday: RAINY'], color='white', fontsize=12, fontweight='bold')

# Pastoryti rėmus
for spine in ax.spines.values():
    spine.set_edgecolor('white')
    spine.set_linewidth(1)

ax.tick_params(colors='white', length=0)

# Pavadinimas
ax.set_title('Markov Transition Matrix: NYC Rainfall\nProbability of tomorrow given today',
             color='white', fontsize=14, fontweight='bold', pad=20)

# Apatinis paaiškinimas
fig.text(0.5, -0.02, 
         f'Calculated from {len(transitions)} consecutive day pairs (2016-2025). '
         'Rainfall shows persistence: rainy days more often follow rainy days than the base rate (36.1%) would suggest.',
         ha='center', color='#a8dadc', fontsize=10, style='italic', wrap=True)

plt.tight_layout()
plt.savefig('markov_matrix.png', dpi=150, bbox_inches='tight', facecolor='#0d1b2a')
plt.show()

print("Matrica išsaugota: markov_matrix.png")
print("\nReikšmės:")
print(transition_matrix.round(3))
```
